```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Load league-wide game data and filter to MIL home games
data = pd.read_csv('../../../data/league_weather_2021_2025.csv')
mil = data[data['home_team'] == 'MIL'].copy()

# Parse game_date to extract month
mil['game_date'] = pd.to_datetime(mil['game_date'])
mil['month'] = mil['game_date'].dt.month
month_map = {3: 'Mar', 4: 'Apr', 5: 'May', 6: 'Jun',
             7: 'Jul', 8: 'Aug', 9: 'Sep', 10: 'Oct'}
mil['month_name'] = mil['month'].map(month_map)

print(f"Total MIL home games: {len(mil)}")
print(f"\nGames per month:")
print(mil.groupby('month_name').size().reindex([month_map[m] for m in sorted(mil['month'].unique())]))
```

```python
# ============================================================
# Average Wind Speed by Month
# ============================================================

months = sorted(mil['month'].unique())
month_names = [month_map.get(m, str(m)) for m in months]

grouped = mil.groupby('month')['wspd_mph']
means = grouped.mean().reindex(months)
sems = grouped.sem().reindex(months)
x = np.arange(len(months))

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(x, means, yerr=sems, capsize=4,
              color='#2ca02c', alpha=0.75, edgecolor='black', linewidth=0.5)
ax.set_xticks(x)
ax.set_xticklabels(month_names, fontsize=11)
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Wind Speed (mph)', fontsize=12)
ax.set_title('Average Wind Speed by Month — American Family Field', fontsize=13)
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)

for xi, m in zip(x, means):
    ax.text(xi, m + sems.iloc[xi] + 0.1, f'{m:.1f}', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Average Pressure by Month
# ============================================================

grouped = mil.groupby('month')['pres']
means = grouped.mean().reindex(months)
sems = grouped.sem().reindex(months)

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(x, means, yerr=sems, capsize=4,
              color='#9467bd', alpha=0.75, edgecolor='black', linewidth=0.5)
ax.set_xticks(x)
ax.set_xticklabels(month_names, fontsize=11)
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Pressure (hPa)', fontsize=12)
ax.set_title('Average Pressure by Month — American Family Field', fontsize=13)
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)

for xi, m in zip(x, means):
    ax.text(xi, m + sems.iloc[xi] + 0.1, f'{m:.1f}', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Summary Statistics Table
# ============================================================

summary = mil.groupby('month').agg(
    games=('wspd_mph', 'size'),
    wspd_mean=('wspd_mph', 'mean'),
    wspd_std=('wspd_mph', 'std'),
    pres_mean=('pres', 'mean'),
    pres_std=('pres', 'std'),
    rhum_mean=('rhum', 'mean'),
    rhum_std=('rhum', 'std'),
).round(2)

summary.index = [month_map.get(m, str(m)) for m in summary.index]
summary.index.name = 'Month'
summary.columns = ['Games', 'Wind Mean (mph)', 'Wind Std',
                    'Pressure Mean (hPa)', 'Pressure Std',
                    'Humidity Mean (%)', 'Humidity Std']
summary
```
