# Weather Across the Season at Rogers Centre (Focus: Pressure)

**Goal:** Visualize how temperature, pressure, humidity, and wind speed vary by month at Rogers Centre (Toronto Blue Jays), with a particular focus on atmospheric pressure and its relationship to game outcomes.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Load league-wide game data and filter to TOR home games
data = pd.read_csv('../../data/league_weather_2021_2025.csv')
tor = data[data['home_team'] == 'TOR'].copy()

# Parse game_date to extract month
tor['game_date'] = pd.to_datetime(tor['game_date'])
tor['month'] = tor['game_date'].dt.month
month_map = {3: 'Mar', 4: 'Apr', 5: 'May', 6: 'Jun',
             7: 'Jul', 8: 'Aug', 9: 'Sep', 10: 'Oct'}
tor['month_name'] = tor['month'].map(month_map)

months = sorted(tor['month'].unique())
month_names = [month_map.get(m, str(m)) for m in months]

print(f"Total TOR home games: {len(tor)}")
print(f"\nGames per month:")
print(tor.groupby('month_name').size().reindex([month_map[m] for m in months]))
```

```python
# ============================================================
# Weather Distributions by Month
# ============================================================

weather_vars = {
    'temp_f':   {'label': 'Temperature (\u00b0F)', 'color': '#d62728'},
    'rhum':     {'label': 'Humidity (%)',      'color': '#1f77b4'},
    'wspd_mph': {'label': 'Wind Speed (mph)',  'color': '#2ca02c'},
    'pres':     {'label': 'Pressure (hPa)',    'color': '#9467bd'},
}

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for ax, (col, info) in zip(axes.flat, weather_vars.items()):
    box_data = [tor.loc[tor['month'] == m, col].dropna().values for m in months]
    bp = ax.boxplot(box_data, patch_artist=True, labels=month_names,
                    medianprops=dict(color='black', linewidth=1.5))
    for patch in bp['boxes']:
        patch.set_facecolor(info['color'])
        patch.set_alpha(0.6)
    ax.set_xlabel('Month', fontsize=11)
    ax.set_ylabel(info['label'], fontsize=11)
    ax.set_title(info['label'], fontsize=13)
    ax.yaxis.grid(True, alpha=0.3)
    ax.set_axisbelow(True)

fig.suptitle('Weather Distributions by Month \u2014 Rogers Centre', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Scatter Plots: Temperature vs. Pressure & Temperature vs. Humidity
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Temperature vs. Pressure
ax = axes[0]
scatter = ax.scatter(tor['temp_f'], tor['pres'], c=tor['month'], cmap='coolwarm',
                     alpha=0.6, edgecolors='black', linewidth=0.3, s=40)
z = np.polyfit(tor['temp_f'].dropna(), tor.loc[tor['temp_f'].notna(), 'pres'], 1)
p = np.poly1d(z)
x_line = np.linspace(tor['temp_f'].min(), tor['temp_f'].max(), 100)
ax.plot(x_line, p(x_line), color='black', linewidth=1.5, linestyle='--')
ax.set_xlabel('Temperature (\u00b0F)', fontsize=11)
ax.set_ylabel('Pressure (hPa)', fontsize=11)
ax.set_title('Temperature vs. Pressure', fontsize=13)
ax.yaxis.grid(True, alpha=0.3)
ax.xaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)

# Temperature vs. Humidity
ax = axes[1]
scatter = ax.scatter(tor['temp_f'], tor['rhum'], c=tor['month'], cmap='coolwarm',
                     alpha=0.6, edgecolors='black', linewidth=0.3, s=40)
z = np.polyfit(tor['temp_f'].dropna(), tor.loc[tor['temp_f'].notna(), 'rhum'], 1)
p = np.poly1d(z)
ax.plot(x_line, p(x_line), color='black', linewidth=1.5, linestyle='--')
ax.set_xlabel('Temperature (\u00b0F)', fontsize=11)
ax.set_ylabel('Humidity (%)', fontsize=11)
ax.set_title('Temperature vs. Humidity', fontsize=13)
ax.yaxis.grid(True, alpha=0.3)
ax.xaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)

cbar = fig.colorbar(scatter, ax=axes, label='Month', ticks=months)
cbar.ax.set_yticklabels(month_names)

fig.suptitle('Weather Relationships \u2014 Rogers Centre', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Day vs. Night classification
# ============================================================

# Day games start before 5 PM, night games at 5 PM or later
tor['time_of_day'] = tor['start_hour'].apply(lambda h: 'Day' if h < 17 else 'Night')

print("Game counts by month and time of day:")
print(tor.groupby(['month_name', 'time_of_day']).size().unstack(fill_value=0)
      .reindex([month_map[m] for m in months]))
```

```python
# ============================================================
# Average Temperature: Day vs. Night Games by Month
# ============================================================

fig, ax = plt.subplots(figsize=(12, 6))
x = np.arange(len(months))
width = 0.35

for i, (label, color) in enumerate([('Day', '#ff9900'), ('Night', '#003366')]):
    subset = tor[tor['time_of_day'] == label]
    grouped = subset.groupby('month')['temp_f']
    means = grouped.mean().reindex(months)
    sems = grouped.sem().reindex(months)
    offset = (i - 0.5) * width
    ax.bar(x + offset, means, width, yerr=sems, capsize=4,
           label=label, color=color, alpha=0.75, edgecolor='black', linewidth=0.5)
    for xi, m in zip(x, means):
        if pd.notna(m):
            ax.text(xi + offset, m + sems.reindex(months).iloc[xi] + 0.3,
                    f'{m:.1f}', ha='center', va='bottom', fontsize=9)

ax.set_xticks(x)
ax.set_xticklabels(month_names, fontsize=11)
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Temperature (\u00b0F)', fontsize=12)
ax.set_title('Average Temperature: Day vs. Night Games by Month \u2014 Rogers Centre', fontsize=13)
ax.legend(fontsize=11)
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Average Humidity: Day vs. Night Games by Month
# ============================================================

fig, ax = plt.subplots(figsize=(12, 6))
x = np.arange(len(months))
width = 0.35

for i, (label, color) in enumerate([('Day', '#ff9900'), ('Night', '#003366')]):
    subset = tor[tor['time_of_day'] == label]
    grouped = subset.groupby('month')['rhum']
    means = grouped.mean().reindex(months)
    sems = grouped.sem().reindex(months)
    offset = (i - 0.5) * width
    ax.bar(x + offset, means, width, yerr=sems, capsize=4,
           label=label, color=color, alpha=0.75, edgecolor='black', linewidth=0.5)
    for xi, m in zip(x, means):
        if pd.notna(m):
            ax.text(xi + offset, m + sems.reindex(months).iloc[xi] + 0.3,
                    f'{m:.1f}', ha='center', va='bottom', fontsize=9)

ax.set_xticks(x)
ax.set_xticklabels(month_names, fontsize=11)
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Humidity (%)', fontsize=12)
ax.set_title('Average Humidity: Day vs. Night Games by Month \u2014 Rogers Centre', fontsize=13)
ax.legend(fontsize=11)
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Average Pressure: Day vs. Night Games by Month
# ============================================================

fig, ax = plt.subplots(figsize=(12, 6))
x = np.arange(len(months))
width = 0.35

all_means = []
all_sems = []
for i, (label, color) in enumerate([('Day', '#ff9900'), ('Night', '#003366')]):
    subset = tor[tor['time_of_day'] == label]
    grouped = subset.groupby('month')['pres']
    means = grouped.mean().reindex(months)
    sems = grouped.sem().reindex(months)
    all_means.append(means)
    all_sems.append(sems)
    offset = (i - 0.5) * width
    ax.bar(x + offset, means, width, yerr=sems, capsize=4,
           label=label, color=color, alpha=0.75, edgecolor='black', linewidth=0.5)
    for xi, m in zip(x, means):
        if pd.notna(m):
            ax.text(xi + offset, m + sems.reindex(months).iloc[xi] + 0.1,
                    f'{m:.1f}', ha='center', va='bottom', fontsize=9)

# Zoom y-axis to better show differences
y_min = min(m.min() for m in all_means) - 2
y_max = max(m.max() + s.max() for m, s in zip(all_means, all_sems)) + 1.5
ax.set_ylim(y_min, y_max)

ax.set_xticks(x)
ax.set_xticklabels(month_names, fontsize=11)
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Pressure (hPa)', fontsize=12)
ax.set_title('Average Pressure: Day vs. Night Games by Month \u2014 Rogers Centre', fontsize=13)
ax.legend(fontsize=11)
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Average Wind Speed: Day vs. Night Games by Month
# ============================================================

fig, ax = plt.subplots(figsize=(12, 6))
x = np.arange(len(months))
width = 0.35

for i, (label, color) in enumerate([('Day', '#ff9900'), ('Night', '#003366')]):
    subset = tor[tor['time_of_day'] == label]
    grouped = subset.groupby('month')['wspd_mph']
    means = grouped.mean().reindex(months)
    sems = grouped.sem().reindex(months)
    offset = (i - 0.5) * width
    ax.bar(x + offset, means, width, yerr=sems, capsize=4,
           label=label, color=color, alpha=0.75, edgecolor='black', linewidth=0.5)
    for xi, m in zip(x, means):
        if pd.notna(m):
            ax.text(xi + offset, m + sems.reindex(months).iloc[xi] + 0.1,
                    f'{m:.1f}', ha='center', va='bottom', fontsize=9)

ax.set_xticks(x)
ax.set_xticklabels(month_names, fontsize=11)
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Wind Speed (mph)', fontsize=12)
ax.set_title('Average Wind Speed: Day vs. Night Games by Month \u2014 Rogers Centre', fontsize=13)
ax.legend(fontsize=11)
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Typical Day Game Wind Direction by Month (Polar Rose)
# ============================================================

def deg_to_compass(deg):
    dirs = ['N', 'NE', 'E', 'SE', 'S', 'SW', 'W', 'NW']
    idx = int((deg + 22.5) % 360 / 45)
    return dirs[idx]

compass_order = ['N', 'NE', 'E', 'SE', 'S', 'SW', 'W', 'NW']
compass_angles = {d: np.radians(i * 45) for i, d in enumerate(compass_order)}

day_games = tor[tor['time_of_day'] == 'Day'].copy()
day_games['compass'] = day_games['wdir'].apply(deg_to_compass)

active_months = sorted(day_games['month'].unique())
n_months = len(active_months)
cols = min(4, n_months)
rows = int(np.ceil(n_months / cols))

fig, axes = plt.subplots(rows, cols, figsize=(4 * cols, 4 * rows),
                         subplot_kw=dict(projection='polar'))
if n_months == 1:
    axes = np.array([axes])
axes = axes.flat

for idx, m in enumerate(active_months):
    ax = axes[idx]
    subset = day_games[day_games['month'] == m]
    counts = subset['compass'].value_counts().reindex(compass_order, fill_value=0)
    angles = [compass_angles[d] for d in compass_order]
    bars = ax.bar(angles, counts, width=np.radians(40), alpha=0.7,
                  color='#2ca02c', edgecolor='black', linewidth=0.5)
    ax.set_theta_zero_location('N')
    ax.set_theta_direction(-1)
    ax.set_thetagrids([i * 45 for i in range(8)], compass_order, fontsize=9)
    ax.set_title(f'{month_map[m]}  (n={len(subset)})', fontsize=11, pad=15)
    ax.set_yticklabels([])

for idx in range(n_months, rows * cols):
    axes[idx].set_visible(False)

fig.suptitle('Day Game Wind Direction by Month \u2014 Rogers Centre', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Typical Night Game Wind Direction by Month (Polar Rose)
# ============================================================

night_games = tor[tor['time_of_day'] == 'Night'].copy()
night_games['compass'] = night_games['wdir'].apply(deg_to_compass)

active_months_n = sorted(night_games['month'].unique())
n_months_n = len(active_months_n)
cols = min(4, n_months_n)
rows = int(np.ceil(n_months_n / cols))

fig, axes = plt.subplots(rows, cols, figsize=(4 * cols, 4 * rows),
                         subplot_kw=dict(projection='polar'))
if n_months_n == 1:
    axes = np.array([axes])
axes = axes.flat

for idx, m in enumerate(active_months_n):
    ax = axes[idx]
    subset = night_games[night_games['month'] == m]
    counts = subset['compass'].value_counts().reindex(compass_order, fill_value=0)
    angles = [compass_angles[d] for d in compass_order]
    bars = ax.bar(angles, counts, width=np.radians(40), alpha=0.7,
                  color='#003366', edgecolor='black', linewidth=0.5)
    ax.set_theta_zero_location('N')
    ax.set_theta_direction(-1)
    ax.set_thetagrids([i * 45 for i in range(8)], compass_order, fontsize=9)
    ax.set_title(f'{month_map[m]}  (n={len(subset)})', fontsize=11, pad=15)
    ax.set_yticklabels([])

for idx in range(n_months_n, rows * cols):
    axes[idx].set_visible(False)

fig.suptitle('Night Game Wind Direction by Month \u2014 Rogers Centre', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Average Total Runs: Day vs. Night Games by Month
# ============================================================

fig, ax = plt.subplots(figsize=(12, 6))
x = np.arange(len(months))
width = 0.35

for i, (label, color) in enumerate([('Day', '#ff9900'), ('Night', '#003366')]):
    subset = tor[tor['time_of_day'] == label]
    grouped = subset.groupby('month')['total_runs']
    means = grouped.mean().reindex(months)
    sems = grouped.sem().reindex(months)
    offset = (i - 0.5) * width
    ax.bar(x + offset, means, width, yerr=sems, capsize=4,
           label=label, color=color, alpha=0.75, edgecolor='black', linewidth=0.5)
    for xi, m in zip(x, means):
        if pd.notna(m):
            ax.text(xi + offset, m + sems.reindex(months).iloc[xi] + 0.1,
                    f'{m:.1f}', ha='center', va='bottom', fontsize=9)

ax.set_xticks(x)
ax.set_xticklabels(month_names, fontsize=11)
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Total Runs', fontsize=12)
ax.set_title('Average Total Runs: Day vs. Night Games by Month \u2014 Rogers Centre', fontsize=13)
ax.legend(fontsize=11)
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Correlations with Pressure: Runs, Home Runs, Strikeouts
# ============================================================

corr_vars = {
    'total_runs':     {'label': 'Total Runs',     'color': '#d62728'},
    'home_runs_hit':  {'label': 'Home Runs',       'color': '#ff7f0e'},
    'strikeouts':     {'label': 'Strikeouts',      'color': '#1f77b4'},
}

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, (col, info) in zip(axes, corr_vars.items()):
    valid = tor.dropna(subset=['pres', col])
    r = valid['pres'].corr(valid[col])
    ax.scatter(valid['pres'], valid[col], alpha=0.5, s=30,
               color=info['color'], edgecolors='black', linewidth=0.3)
    # Trend line
    z = np.polyfit(valid['pres'], valid[col], 1)
    p = np.poly1d(z)
    x_line = np.linspace(valid['pres'].min(), valid['pres'].max(), 100)
    ax.plot(x_line, p(x_line), color='black', linewidth=1.5, linestyle='--')
    ax.set_xlabel('Pressure (hPa)', fontsize=11)
    ax.set_ylabel(info['label'], fontsize=11)
    ax.set_title(f'{info["label"]} vs. Pressure (r = {r:.3f})', fontsize=13)
    ax.xaxis.grid(True, alpha=0.3)
    ax.yaxis.grid(True, alpha=0.3)
    ax.set_axisbelow(True)

fig.suptitle('Correlations with Pressure \u2014 Rogers Centre', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

# Print correlation summary
print("Pearson correlations with pressure:")
for col, info in corr_vars.items():
    valid = tor.dropna(subset=['pres', col])
    r = valid['pres'].corr(valid[col])
    print(f"  {info['label']:15s}: r = {r:+.4f}  (n = {len(valid)})")
```

## Pressure Deep Dive

The following cells explore atmospheric pressure at Rogers Centre in greater detail: how it distributes across the season, how high- vs. low-pressure games differ in scoring, and how pressure interacts with other weather variables.

```python
# ============================================================
# Pressure Distribution: Histogram + KDE by Month
# ============================================================

fig, ax = plt.subplots(figsize=(14, 6))
colors = plt.cm.coolwarm(np.linspace(0.1, 0.9, len(months)))

for m, color in zip(months, colors):
    subset = tor.loc[tor['month'] == m, 'pres'].dropna()
    ax.hist(subset, bins=15, alpha=0.35, color=color, label=month_map[m],
            edgecolor='black', linewidth=0.3)

ax.axvline(tor['pres'].median(), color='black', linestyle='--', linewidth=1.5,
           label=f'Overall Median ({tor["pres"].median():.1f} hPa)')
ax.set_xlabel('Pressure (hPa)', fontsize=12)
ax.set_ylabel('Game Count', fontsize=12)
ax.set_title('Pressure Distribution by Month \u2014 Rogers Centre', fontsize=14)
ax.legend(fontsize=10)
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# High vs. Low Pressure Game Outcomes
# ============================================================

median_pres = tor['pres'].median()
tor['pres_group'] = tor['pres'].apply(lambda x: 'High' if x >= median_pres else 'Low')

outcome_vars = ['total_runs', 'home_runs_hit', 'strikeouts']
outcome_labels = ['Total Runs', 'Home Runs', 'Strikeouts']

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

for ax, col, label in zip(axes, outcome_vars, outcome_labels):
    high = tor.loc[tor['pres_group'] == 'High', col].dropna()
    low = tor.loc[tor['pres_group'] == 'Low', col].dropna()

    bp = ax.boxplot([low.values, high.values], patch_artist=True,
                    labels=[f'Low\n(<{median_pres:.0f} hPa)\nn={len(low)}',
                            f'High\n(\u2265{median_pres:.0f} hPa)\nn={len(high)}'],
                    medianprops=dict(color='black', linewidth=2))
    bp['boxes'][0].set_facecolor('#ff6b6b')
    bp['boxes'][0].set_alpha(0.7)
    bp['boxes'][1].set_facecolor('#4dabf7')
    bp['boxes'][1].set_alpha(0.7)

    ax.set_ylabel(label, fontsize=12)
    ax.set_title(f'{label}: Low vs. High Pressure', fontsize=13)
    ax.yaxis.grid(True, alpha=0.3)
    ax.set_axisbelow(True)

    # Annotate means
    for i, (vals, lbl) in enumerate([(low, 'Low'), (high, 'High')]):
        ax.text(i + 1, vals.mean(), f'\u03bc={vals.mean():.2f}',
                ha='center', va='bottom', fontsize=10, fontweight='bold')

fig.suptitle(f'Game Outcomes by Pressure Group (split at median {median_pres:.0f} hPa) \u2014 Rogers Centre',
             fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

# Print means
print(f"Median pressure split: {median_pres:.1f} hPa\n")
for col, label in zip(outcome_vars, outcome_labels):
    high_mean = tor.loc[tor['pres_group'] == 'High', col].mean()
    low_mean = tor.loc[tor['pres_group'] == 'Low', col].mean()
    print(f"  {label:15s}:  Low = {low_mean:.2f}   High = {high_mean:.2f}   Diff = {high_mean - low_mean:+.2f}")
```

```python
# ============================================================
# Pressure vs. All Weather Variables (Pairwise Scatter Matrix)
# ============================================================

pair_vars = {
    'temp_f':   'Temperature (\u00b0F)',
    'rhum':     'Humidity (%)',
    'wspd_mph': 'Wind Speed (mph)',
}

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, (col, label) in zip(axes, pair_vars.items()):
    valid = tor.dropna(subset=['pres', col])
    r = valid['pres'].corr(valid[col])
    ax.scatter(valid['pres'], valid[col], alpha=0.5, s=30,
               c=valid['month'], cmap='coolwarm', edgecolors='black', linewidth=0.3)
    z = np.polyfit(valid['pres'], valid[col], 1)
    p = np.poly1d(z)
    x_line = np.linspace(valid['pres'].min(), valid['pres'].max(), 100)
    ax.plot(x_line, p(x_line), color='black', linewidth=1.5, linestyle='--')
    ax.set_xlabel('Pressure (hPa)', fontsize=11)
    ax.set_ylabel(label, fontsize=11)
    ax.set_title(f'{label} vs. Pressure (r = {r:.3f})', fontsize=13)
    ax.xaxis.grid(True, alpha=0.3)
    ax.yaxis.grid(True, alpha=0.3)
    ax.set_axisbelow(True)

fig.suptitle('Pressure vs. Other Weather Variables \u2014 Rogers Centre', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

print("Pearson correlations with pressure:")
for col, label in pair_vars.items():
    valid = tor.dropna(subset=['pres', col])
    r = valid['pres'].corr(valid[col])
    print(f"  {label:22s}: r = {r:+.4f}  (n = {len(valid)})")
```

```python
# ============================================================
# Pressure Trend Across the Season (Game-by-Game)
# ============================================================

fig, ax = plt.subplots(figsize=(14, 5))
tor_sorted = tor.sort_values('game_date')

ax.plot(tor_sorted['game_date'], tor_sorted['pres'], color='#9467bd',
        alpha=0.4, linewidth=0.8, marker='o', markersize=3)

# Rolling average (10-game window)
rolling = tor_sorted.set_index('game_date')['pres'].rolling(window=10, min_periods=3).mean()
ax.plot(rolling.index, rolling.values, color='black', linewidth=2, label='10-game rolling avg')

ax.set_xlabel('Date', fontsize=12)
ax.set_ylabel('Pressure (hPa)', fontsize=12)
ax.set_title('Pressure Trend Across the Season \u2014 Rogers Centre', fontsize=14)
ax.legend(fontsize=11)
ax.yaxis.grid(True, alpha=0.3)
ax.xaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
